# Environment Check

This notebook verifies the Python environment and core geospatial packages used in the project.

```python
import sys
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib

print("Python:", sys.version.split()[0])
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("GeoPandas:", gpd.__version__)
print("Matplotlib:", matplotlib.__version__)
```

```python
locations = pd.DataFrame({
    "name": ['Inwood', 'Columbia University', 'Upper East Side','Midtown',
    'Lower Manhattan'],
    "Latitude": [40.8677,
        40.8075,
        40.7736,
        40.7580,
        40.7127],
    "Longitude": [-73.9212,
        -73.9626,
        -73.9566,
        -73.9855,
        -74.0134]
})
gdf = gpd.GeoDataFrame(locations, geometry=gpd.points_from_xy(
    locations['Longitude'], locations['Latitude']),
    crs="EPSG:4326"
)

gdf

```

```python
import matplotlib.pyplot as plt

ax = gdf.plot(
    figsize=(6, 8),
    color="navy",
    markersize=60
)

for x, y, label in zip(
    gdf.geometry.x,
    gdf.geometry.y,
    gdf["name"]
):
    ax.annotate(
        label,
        (x, y),
        xytext=(5, 5),
        textcoords="offset points"
    )

ax.set_title("Sample Manhattan Locations")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")

plt.show()
```
